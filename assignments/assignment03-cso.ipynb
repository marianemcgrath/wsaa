A program that **retrieves the dataset for the "exchequer account (historical series)" from the CSO,
and stores it into a file called "cso.json"**

```python
# Import necessary libraries
import requests
import json

# Define the URL
url = "https://ws.cso.ie/public/api.restful/PxStat.Data.Cube_API.ReadDataset/FIQ02/JSON-stat/2.0/en"
```

```python
# Fetch data from the URL
try:
    response = requests.get(url)
    response.raise_for_status()

    data = response.json()

    with open('cso.json', 'w') as file:    
        json.dump(data, file, indent=4)  # indentation to read better

    print("Data successfully saved to cso.json")

except requests.exceptions.RequestException as e:
    print(f"Error fetching data: {e}")

# Source: https://stackoverflow.com/questions/62418874/using-json-files-to-store-and-retrieve-list-data
```

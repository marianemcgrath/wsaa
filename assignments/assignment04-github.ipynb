A program in python that will read a file from this repository, and it replaces all the instances of the text "Andrew" with "Mariane". 

The program should then commit those changes and push the file back to the repository.

*API Keys must be hidden*

```python
# Import necessary libraries
import os
import base64
import requests

# Get the GitHub token
GITHUB_TOKEN = os.environ.get("GITHUB_TOKEN")
if not GITHUB_TOKEN:
    print("Please set PFDA_GITHUB_TOKEN environment variable.")
    raise Exception("Token not set")

# Define repository and file details
REPO_OWNER = "marianemcgrath"
REPO_NAME = "pfda"
FILE_PATH = "andrew_file.txt"
BRANCH = "main"

# Construct the API URL
API_URL = f"https://api.github.com/repos/{REPO_OWNER}/{REPO_NAME}/contents/{FILE_PATH}"

# Set up headers for authentication
HEADERS = {"Authorization": f"Bearer {GITHUB_TOKEN}",
    "Accept": "application/vnd.github+json"}
```

```python
# Get the file original content and sha 
print("Fetching file...")
response = requests.get(API_URL, headers=HEADERS, params={"ref": BRANCH})
response.raise_for_status()

data = response.json()

content = base64.b64decode(data["content"]).decode("utf-8")
sha = data["sha"]   # The SHA is needed to update the file

print("Original content:")
print(content)
```

```python
# Changing the content (the name "Andrew" to "Mariane")
updated_content = content.replace("Andrew", "Mariane")

if updated_content == content:
    print("No changes needed.")
else:
    print("Updating file...")

    encoded = base64.b64encode(updated_content.encode("utf-8")).decode("utf-8")

    payload = {"message": "Replace Andrew with Mariane",
        "content": encoded, "sha": sha, "branch": BRANCH}
```

```python
# Pushing changes to repo
push = requests.put(API_URL, headers=HEADERS, json=payload)
    
print(push.status_code)
print(push.json())
```

A program in python that will read a file from this repository, and replace all the instances of the text "Andrew" with "Mariane". 

The program should then commit those changes and push the file back to the repository. NOTE: *API Keys must be hidden*

```python
# Import necessary libraries
import os
import base64   # for encoding and decoding content to/from base64
import requests

# Get the GitHub token
GITHUB_TOKEN = os.environ.get("GITHUB_TOKEN")
if not GITHUB_TOKEN:
    print("Please set GITHUB_TOKEN environment variable.")
    raise Exception("Token not set")

# Define repository and file details
REPO_OWNER = "marianemcgrath"
REPO_NAME = "pfda"
FILE_PATH = "andrew_file.txt"
BRANCH = "main"

# Build the API URL
API_URL = f"https://api.github.com/repos/{REPO_OWNER}/{REPO_NAME}/contents/{FILE_PATH}"

# Set up headers for authentication
HEADERS = {"Authorization": f"Bearer {GITHUB_TOKEN}",
    "Accept": "application/vnd.github+json"}  # recommended by GitHub docs

# Source: https://docs.github.com/en/rest/authentication/authenticating-to-the-rest-api?apiVersion=2026-03-10 (authenticating to GitHub API)
# Source: https://docs.python.org/3/library/base64.html (base64 encoding and decoding)

```

```python
# Get the file original content and sha 
print("Fetching file...")

response = requests.get(API_URL, headers=HEADERS, params={"ref": BRANCH})
response.raise_for_status()

data = response.json()

# Decode the content from base64 and print it
content = base64.b64decode(data["content"].replace("\n", "")).decode("utf-8")
sha = data["sha"]   # The SHA is needed to update the file

# Source: https://www.datacamp.com/tutorial/making-http-requests-in-python?utm_cid=23340058065&utm_aid=192632748929&utm_campaign=230119_1-ps-dscia~dsa-tofu~python_2-b2c_3-emea_4-prc_5-na_6-na_7-le_8-pdsh-go_9-nb-e_10-na_11-na&utm_loc=1007846-&utm_mtd=-c&utm_kw=&utm_source=google&utm_medium=paid_search&utm_content=ps-dscia~emea-en~dsa~tofu~tutorial~python&gad_source=1&gad_campaignid=23340058065&gbraid=0AAAAADQ9WsFyBfVwMhyNHZD5vcwboFtIY&gclid=CjwKCAjw1tLOBhAMEiwAiPkRHnK8hkJWEWqwYBZk6k7RwzBwobX3L4OBis7BSv7EQ2ntM7noigvNABoC2_oQAvD_BwE (HTTP requests in Python)
# Source: https://python.plainenglish.io/understand-encoding-and-hashing-in-python-904874815872 (understanding SHA)
```

```python
# Print the original content
print("Original content:")
print(content)

# Changing the content (the name "Andrew" to "Mariane")
updated_content = content.replace("Andrew", "Mariane")

if updated_content == content:
    print("No changes needed.")
else:
    print("Updating file...")

    encoded = base64.b64encode(updated_content.encode("utf-8")).decode("utf-8")

    payload = {
        "message": "Replace Andrew with Mariane",
        "content": encoded,
        "sha": sha,
        "branch": BRANCH
        }
    
    # Pushing changes back to the repo
    response = requests.put(API_URL, headers=HEADERS, json=payload)
    response.raise_for_status()

    print("File updated successfully.")

# Text source: https://en.wikipedia.org/wiki/Andrew_Ventura
# Source: https://requests.readthedocs.io/en/latest/user/quickstart/ (HTTP requests in Python, several sections)
```
